```python
import numpy as np
import panda as pd
```

```python
# Membuat berapa banyak data yang ingin di latih
jumlah_data = 1000
```

```python
jenis_hewan = np.random.choice(['Kambing', 'Domba'], jumlah_data)
umur_bulan = np.random.randint(3, 48, jumlah_data)
berat_aktual = [10 + (umur * 1.2) + np.random.normal(0,3) for umur in umur_bulan]
```

```python
# Label status fisiologis
status = np.random.choice(['Pertumbuhan', 'Dewasa Maintanance', 'Hamil', 'Menyusui'], jumlah_data)
```

```python
# Menghitung target pakan berdasrakan rumus standar NRC (National Research Council)
# (Dewan Penelitian Nasional Amerika Serikat) National Academies.
# Dummy logic untuk membuat target hijauan & konsentrat

target_hijauan = [(b * 0.03) * 0.6 for b in berat_aktual] # 60% dari kebutuhan pakan
target_konsentrat = [(b * 0.03) * 0.4 for b in berat_aktual] # 40% dari kebutuhan pakan
```

```python
print("Jenis Hewan:", len(jenis_hewan))
print("Umur Bulan:", len(umur_bulan))
print("Berat Aktual:", len(berat_aktual))
print("Status Fisiologis:", len(status))
print("Target Hijauan:", len(target_hijauan))
print("Target Konsentrat:", len(target_konsentrat))
```

```python
# Menyimpan ke File CSV
data = {
    'jenis_hewan': jenis_hewan,
    'umur_bulan': umur_bulan,
    'berat_aktual': berat_aktual,
    'status_fisiologis': status,
    'target_hijauan': target_hijauan,
    'target_konsentrat': target_konsentrat
}
df = pd.DataFrame(data)

df.to_csv('data_ternak.csv', index=False)
print("Berhasil! File 'data_ternak.csv' sudah dibuat di folder kamu.")
```
